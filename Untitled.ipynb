```python
import matplotlib.pyplot as plt

# Real data
# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

# Plot
plt.figure(figsize=(10, 5))
plt.plot(years, oscar_revenue, marker='o')

plt.title("Oscar Winning Movies Revenue")
plt.xlabel("oscar_movie")
plt.ylabel("Revenue (Million USD)")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]
non_oscar_revenue = [378, 2788, 829, 185, 275]

x = np.arange(len(years))   # positions
width = 0.35                # bar width

plt.style.use("ggplot")
plt.figure(figsize=(10, 5))

plt.bar(x - width/2, oscar_revenue, width, label="Oscar Winners")
plt.bar(x + width/2, non_oscar_revenue, width, label="Non-Oscar Winners")

plt.xticks(x, years)
plt.xlabel("Year")
plt.ylabel("Revenue (Million USD)")
plt.title("Oscar vs Non-Oscar Winning Movies Revenue")
plt.legend()
plt.grid(True)

plt.show()

```

```python
import matplotlib.pyplot as plt

# Game of the Year data
goty_games = [
    "Sekiro: Shadows Die Twice",
    "The Last of Us Part II",
    "It Takes Two",
    "Elden Ring",
    "Baldur's Gate 3",
    "Alan Wake 2",
    "Final Fantasy VII Rebirth"
]

goty_revenue = [500, 1000, 300, 2500, 1200, 350, 900]  # in $M

plt.figure(figsize=(12, 6))
plt.bar(goty_games, goty_revenue)

plt.title("Game of the Year Winners – Estimated Revenue")
plt.xlabel("Game")
plt.ylabel("Revenue (Million USD)")

plt.xticks(rotation=5, ha="right")

# Annotate bars
for game, revenue in zip(goty_games, goty_revenue):
    plt.text(
        game,
        revenue,
        str(revenue),
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Stack Plot
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

direct = [50, 60, 70, 80, 90, 100, 110]
organic = [30, 40, 50, 55, 60, 70, 80]
social = [20, 25, 30, 35, 40, 50, 60]

plt.stackplot(
    days,
    direct,
    organic,
    social,
    labels=["Direct", "Organic", "Social"]
)

plt.xlabel("Days")
plt.ylabel("Traffic")
plt.title("Website Traffic Sources Over a Week")
plt.legend(loc="upper left")
plt.show()
```

```python
# Task
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25],  # New York
    [18, 19, 17, 20, 21],  # London
    [30, 32, 31, 33, 34],  # Delhi
    [25, 26, 24, 27, 28]   # Tokyo
]
fig,axes = plt.subplots(2,2)
city_no = 0
for i in range(2): #rows
    for j in range(2): #columns
        axes[i][j].plot(days,temperatures[city_no])
        axes[i][j].set_title(cities[city_no])
        axes[i][j].grid(True)
        
        city_no +=1
fig.suptitle("Temperature in cities over a week")
fig.tight_layout()
```
